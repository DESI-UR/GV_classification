```python
from astropy.table import Table, vstack, hstack
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
import matplotlib
import matplotlib as mpl

from IPython.display import Image

```

```python
matplotlib.rcParams['figure.dpi'] = 100
matplotlib.rcParams['savefig.dpi'] = 300
```

## load data

```python
DATA_FOLDER = '/pscratch/sd/n/nravi/GV_classification/'
```

```python
plot_dir = DATA_FOLDER + 'sdss_paper_figures/'
```

```python
kias = Table.read(DATA_FOLDER + 'NSA_v1_0_1_vflag_Portsmouth.fits')
```

```python
nsa = Table.read(DATA_FOLDER + 'nsa_v1_0_1_cd_v4_cinv_kias.fits')
```

```python
nsa[:5]
```

```python
nsa_dict = {}
for i in range(len(nsa)):

    nsa_dict[nsa['IAUNAME'][i]] = i
```

```python
nsa['NSA_u_r'] = nsa['ELPETRO_ABSMAG'][:,2]-nsa['ELPETRO_ABSMAG'][:,4]
nsa['NSA_cinv'] = nsa['ELPETRO_TH50'][:,5]/nsa['ELPETRO_TH90'][:,5]
```

```python
kias[:5]
```

```python
nsa['KIAS_u_r'] = np.ones(len(nsa))*np.nan
for i in range(len(kias)):

    idx = nsa_dict[kias['IAUNAME'][i]]

    nsa['KIAS_u_r'][idx] = kias['u_r_KIAS'][i]
```

## Fig 1a color vs magnitude

```python
h, xs, ys, im = plt.hist2d(nsa['ELPETRO_ABSMAG'][:,4],nsa['ELPETRO_ABSMAG'][:,2]-nsa['ELPETRO_ABSMAG'][:,4],
                           bins=(100,100),
                          range=((-24,-12),(0,5)), cmap='Greys',
                          norm=mpl.colors.LogNorm())

plt.colorbar()
```

```python
fig, ax = plt.subplots()
plt.contourf(h.transpose(), extent=[xs.min(), xs.max(), ys.min(), ys.max()], alpha=1,cmap='Greys',
            levels=np.logspace(-1,np.log10(1700),17),
            norm=mpl.colors.LogNorm(),
             # levels=14,
             # edgecolor='none'
            )
plt.gca().invert_xaxis()

# plt.colorbar()
plt.ylabel(r'$u-r$', fontsize=18)
plt.xlabel(r'$M_r$', fontsize=18)
plt.tick_params(labelsize=16, direction='in', axis='both')

# for axis in ['bottom','left']:
#     ax.spines[axis].set_linewidth(3)

# ax.spines['top'].set_alpha(0)
# ax.spines['right'].set_alpha(0)

plt.savefig(plot_dir + 'urNSA_CMD_contour.png', bbox_inches='tight')
```

## Fig 2a -- color vs color gradient
